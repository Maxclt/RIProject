# Compare Python and Fortran Speed

## Import

### Python Packages

```python
import ctypes
import numpy as np
import time
import os
import matplotlib.pyplot as plt
```

### Fortran Lib

```python
lib = ctypes.CDLL("./build/libfortran_code.so")
```

### Import Fortran function

```python
lib.add_numbers.argtypes = [ctypes.c_double, ctypes.c_double]
lib.add_numbers.restype = ctypes.c_double

lib.sum_of_powers.argtypes = [
    np.ctypeslib.ndpointer(dtype=np.float64, flags="C_CONTIGUOUS"),
    ctypes.c_int,
    ctypes.c_int,
]
lib.sum_of_powers.restype = ctypes.c_double

lib.parallelized_sum_of_powers.argtypes = [
    np.ctypeslib.ndpointer(dtype=np.float64, flags="C_CONTIGUOUS"),
    ctypes.c_int,
    ctypes.c_int,
]
lib.parallelized_sum_of_powers.restype = ctypes.c_double
```

## Fortran Test

```python
a = 5
b = 2.5
result = lib.add_numbers(a, b)

print(f"{a} + {b} = {result}")
```

## Define Speed Test

```python
N = 100000000  # 10 million elements
arr = np.array(list(range(N)), dtype=np.float64)
power = 2

def python_sum_of_powers(array, pow):
    res = 0.
    for i in range(len(array)):
        res += array[i] ** pow
    return res
```

### Python Speed

```python
start_py = time.time()
result_py = python_sum_of_powers(arr, power)
end_py = time.time()

print(f"Python result: {result_py:.6f}")
print(f"Python execution time: {end_py - start_py:.6f} seconds")
```

### Fortran Speed

```python
start_ft = time.time()
result_ft = lib.sum_of_powers(arr, N, power)
end_ft = time.time()

print(f"Fortran result: {result_ft:.6f}")
print(f"Fortran execution time: {end_ft - start_ft:.6f} seconds")
```

### Parallelized Fortran Speed

```python
start_ft = time.time()
result_ft = lib.parallelized_sum_of_powers(arr, N, power)
end_ft = time.time()

print(f"Parallelized Fortran result: {result_ft:.6f}")
print(f"Parallelized Fortran execution time: {end_ft - start_ft:.6f} seconds")
```
